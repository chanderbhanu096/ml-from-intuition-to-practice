Each task has a **stub cell** for your answer and a **check cell** that tests it. Write your

Three rules for making this diagnostic honest:

1. **No looking things up.** Not the internet, not the solutions, not your old code. Looking
   things up is a completely legitimate way to work and a completely useless way to measure what
   you know cold.
2. **Give each task about ninety seconds.** If you have not started writing after ninety
   seconds, that is a "no" - mark it and move on.
3. **A wrong answer you nearly had still counts as a "no".** You are choosing what to study, not
   sitting an exam. Being generous with yourself here costs you time later.

Run the setup cell, then work down. Unfilled tasks report "not attempted" rather than failing,
so the notebook runs cleanly either way.

```python
import numpy as np
import pandas as pd

score = {}

def check(task, answer, expected, note=""):
    """Compare an answer with what was expected and record the result."""
    if answer is None:
        score[task] = None
        print(f"task {task}: not attempted")
        return
    try:
        if isinstance(expected, (pd.DataFrame, pd.Series)):
            ok = expected.equals(answer if isinstance(answer, type(expected)) else expected.__class__(answer))
        elif isinstance(expected, np.ndarray):
            ok = np.allclose(np.asarray(answer), expected)
        elif isinstance(expected, float):
            ok = abs(float(answer) - expected) < 1e-6
        else:
            ok = answer == expected
    except Exception as exc:
        ok = False
        note = f"{note} (comparison raised {type(exc).__name__})"
    score[task] = bool(ok)
    print(f"task {task}: {'PASS' if ok else 'FAIL'}{'  - ' + note if note and not ok else ''}")

print("ready - work through the ten tasks below")
```
