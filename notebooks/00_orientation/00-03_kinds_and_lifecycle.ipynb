## Why this matters

"Machine learning" is not one thing. It is a family of quite different activities that share a
toolbox, and the first question a practitioner answers about a new task is *which kind is
this?* - because the kind determines what data you need, what a good result even looks like,
and whether your result means anything at all.

Get it wrong in the usual direction and you produce something worse than nothing: a set of
clusters that look like discovered truth, are named confidently in a slide, and are actually an
artefact of a number you chose. This chapter shows that happening, deliberately, on data whose
real structure we know.

You also get the map of the whole workflow, so that when module 02 starts spending three
chapters on data quality you know why.

## The families

The whole taxonomy comes from two questions: **do you have the answers?** and **where do they
come from?**

### Supervised learning - you have the answers

You have examples where the outcome is recorded, and you want to predict that outcome for new
cases. This is most of applied machine learning and most of this course.

It splits by what kind of thing the target is:

- **Regression** - the target is a **number** on a scale, where "close" means something.
  *How many bikes tomorrow?* Errors are measured in the target's units (bikes). Module 05.
- **Classification** - the target is a **category** from a fixed list. *Is this bike damaged?*
  Errors are counted, not measured, and different mistakes usually cost different amounts.
  Module 06.

A useful test when you are unsure: **is 7 nearer to 8 than to 2?** If yes, it is a number and
you are doing regression. Postcodes are digits and are not numbers in this sense.

### Unsupervised learning - you have no answers

No outcome column exists. You are looking for structure: groups of similar rows (**clustering**),
a smaller set of dimensions that carries most of the information (**dimensionality reduction**),
or rows that do not resemble the rest (**anomaly detection**). Module 08.

The crucial property, and the one people forget: **there is no right answer to check against.**
You cannot validate a clustering the way you validate a prediction, because nothing tells you
what the groups should have been. The failure lab below is about exactly this.

### Semi-supervised learning - a few answers, mostly not

Maria's 400 photos with 30 labelled. Labels are expensive (a person must look), unlabelled data
is cheap. Semi-supervised methods try to use the shape of the unlabelled data to get more out of
the few labels you have. **Active learning** is the neighbouring idea: instead of labelling
randomly, the model asks which 30 photos would teach it the most. Both in module 12-01.

### Self-supervised learning - you manufacture the answers

There is no label, so you invent a task where the label is already in the data: hide part of
each row and predict it from the rest. Nobody had to write anything down, and you can generate
millions of training examples for free.

This is how modern language models are built - hide the next word, predict it, repeat over the
internet - and the point is not the hidden-word task itself but the *representation* the model
is forced to build in order to succeed at it. You will do a small version of this in a moment,
on Maria's data. Modules 10-12 and 12-03.

### Transfer learning - someone else's answers

Start from a model trained on a large related dataset (Hamburg, or a million photographs) and
adapt it with your small dataset. You inherit general structure and only have to learn what is
specific to you. It is the reason a beginner can build a working image classifier on a laptop.
Modules 10-12 and 11-05.

### Reinforcement learning - the answers come from acting

There is no dataset of correct decisions. You act, the world responds with a reward, and you
improve the policy that chooses actions. Maria's weekly price experiment is the simplest
version, a **multi-armed bandit**. It differs from everything above in one deep way: **your
actions change the data you get next**, so exploring and exploiting trade off against each
other. Module 12-05.

### The decision, as a flow

```
Do you have recorded outcomes for the thing you care about?
|
+-- YES, for most rows ........... SUPERVISED
|      target is a number ........... regression      (module 05)
|      target is a category ......... classification  (module 06)
|
+-- YES, for a few rows .......... SEMI-SUPERVISED / ACTIVE LEARNING   (12-01)
|
+-- YES, but for a different
|   but related problem .......... TRANSFER LEARNING  (10-12, 11-05)
|
+-- NO, and I can make some
|   by hiding part of the data ... SELF-SUPERVISED    (10-12, 12-03)
|
+-- NO, and I want structure ..... UNSUPERVISED       (module 08)
|      groups of similar rows ....... clustering
|      fewer dimensions ............. dimensionality reduction
|      unusual rows ................. anomaly detection
|
+-- NO, and the answers only
    exist once I act ............. REINFORCEMENT LEARNING / BANDITS   (12-05)
```

Notice what is *not* in this flow: the size of your data, the industry you work in, and which
library is popular. Those change how you do it, never which one it is.

## One table, three kinds of learning

Rather than three toy examples, here is the same dataset three times. It is the 60 synthetic
days from chapter 00-01 - same seed, same numbers, so you already know the truth behind it:
warm days are busier, rain costs about 22 rentals, and the rest is noise.

**SYNTHETIC**, as before. Nothing here is a fact about any real city.

````python
import numpy as np

rng = np.random.default_rng(1)
n_days = 60
temp_c = rng.integers(10, 29, size=n_days)
is_rainy = rng.random(n_days) < 0.40
rentals = 1.5 * temp_c - 22 * is_rainy + rng.normal(0, 3, n_days)

season = pd.DataFrame({
    "day": np.arange(1, n_days + 1),
    "temp_c": temp_c,
    "weather": np.where(is_rainy, "rainy", "sunny"),
    "rentals": np.clip(np.round(rentals), 0, None).astype(int),
})
train, test = season.iloc[:40], season.iloc[40:]
season.head(3)
````

````python
# 1. SUPERVISED (regression): the target is 'rentals', and we have it for past days.
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error

def encode(rows, columns):
    """Turn the text column 'weather' into a 0/1 column so a model can use it."""
    return pd.get_dummies(rows[columns], columns=["weather"], drop_first=True)

X_train, X_test = encode(train, ["temp_c", "weather"]), encode(test, ["temp_c", "weather"])
supervised = LinearRegression().fit(X_train, train["rentals"])

print("MAE predicting rentals  :", round(mean_absolute_error(test["rentals"], supervised.predict(X_test)), 2), "bikes")
print("MAE of the mean baseline:", round(mean_absolute_error(test["rentals"], [train["rentals"].mean()] * len(test)), 2), "bikes")
````

**1.82 bikes against a baseline of 12.56.** This is the supervised case working exactly as
advertised: an outcome column exists, so we can measure how wrong we are and improve against it.

Now remove the outcome column and see what changes.

````python
# 2. SELF-SUPERVISED: no outcome was ever recorded, so we invent one by hiding a column.
#    Task: predict the temperature from the rest of the row.
Xs_train, Xs_test = encode(train, ["rentals", "weather"]), encode(test, ["rentals", "weather"])
selfsup = LinearRegression().fit(Xs_train, train["temp_c"])

print("MAE predicting temperature:", round(mean_absolute_error(test["temp_c"], selfsup.predict(Xs_test)), 2), "°C")
print("MAE of the mean baseline  :", round(mean_absolute_error(test["temp_c"], [train["temp_c"].mean()] * len(test)), 2), "°C")
````

Nobody labelled anything, and yet there was a training signal: **1.23 °C against a baseline of
5.37 °C.** The label was already inside each row - we simply covered part of it up.

That is the whole idea of self-supervision, and it is worth sitting with for a second, because
the scaled-up version is the most consequential thing in the field right now. Hide the next word
in a sentence and predict it; hide a patch of an image and reconstruct it. The manufactured task
is never the point. The point is that a model cannot do it well without building an internal
representation of how the parts of the data relate - and *that representation* is what gets
reused, through transfer learning, for the task you actually cared about.

**Where the analogy to our toy stops:** predicting temperature from rentals is a genuine
self-supervised task, but it is a single, easy relationship in four columns. Real self-supervised
pre-training gets its power from enormous quantity and variety, which is what makes the learned
representation general rather than a restatement of one correlation. Chapters 10-12 and 12-03
handle it properly, and neither pretends a small notebook creates mastery.

````python
# 3. UNSUPERVISED: forget that 'rentals' or 'weather' mean anything. Just look for groups.
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

Z = StandardScaler().fit_transform(season[["temp_c", "rentals"]])   # scale first: 08-02 explains why
two_groups = KMeans(n_clusters=2, n_init=10, random_state=0).fit(Z)

summary = season.assign(cluster=two_groups.labels_).groupby("cluster").agg(
    days=("day", "size"), mean_temp=("temp_c", "mean"),
    mean_rentals=("rentals", "mean"), share_rainy=("weather", lambda s: (s == "rainy").mean()))
summary.round(2)
````

### Reading the clusters honestly

Two groups of thirty days. Group 0 averages 23.5 °C and 30 rentals; group 1 averages 14.3 °C and
12 rentals. It is very tempting to name them - *"sunny days"* and *"rainy days"* - and write that
on a slide.

Look at the last column before you do. Group 0 is **20% rainy**; group 1 is **47% rainy**. The
direction is right - the busier group does contain fewer wet days - but more than half of the
"rainy cluster" is sunny, and a fifth of the "sunny cluster" is wet. The algorithm never saw the
weather column. It grouped days by temperature and rental level, which blends weather and
temperature into a single "quiet versus busy" axis. **The clusters are real groupings of the two
numbers we handed it, and they are not the categories we would have chosen.**

And notice what is missing compared with the supervised run: there is no MAE, no baseline, no
held-out score. There is nothing to be right or wrong about. That absence is not a gap in this
notebook - it is the defining property of unsupervised learning, and it is why the next section
is a failure lab.

## Failure lab: the segments that were not there

Here is a thing that happens in real companies most weeks.

Someone clusters the customers, gets four groups, gives them names - *Bargain Hunters, Loyal
Regulars, Weekend Browsers, Lapsed* - and the names go into a strategy deck. Six months later
the business is organised around four categories that were produced by typing `n_clusters=4`.

We are going to do it on purpose, with data whose true structure we know: there are exactly two
weather types and one continuous temperature.

**Predict before running:** we will ask for 2, 3, 4 and 5 groups from the same 60 days. How many
of those runs will produce groups that look convincing and could be given names?

````python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 2, figsize=(9, 6.5), sharex=True, sharey=True)
palette = ["#0072B2", "#D55E00", "#009E73", "#CC79A7", "#666666"]

for ax, k in zip(axes.ravel(), [2, 3, 4, 5]):
    labels = KMeans(n_clusters=k, n_init=10, random_state=0).fit(Z).labels_
    for c in range(k):
        m = labels == c
        ax.scatter(season.loc[m, "temp_c"], season.loc[m, "rentals"],
                   color=palette[c], s=40, label=f"group {c}")
    ax.set_title(f"n_clusters = {k}")
    ax.legend(fontsize=7, loc="upper left")

for ax in axes[1]:
    ax.set_xlabel("Temperature (°C)")
for ax in axes[:, 0]:
    ax.set_ylabel("Bikes rented (count)")
fig.suptitle("The same 60 days, four different 'discoveries'")
fig.tight_layout()
plt.show()
````

````python
# What each choice of k would let you claim, in a table.
for k in (2, 3, 4, 5):
    labels = KMeans(n_clusters=k, n_init=10, random_state=0).fit(Z).labels_
    shares = [float((season["weather"][labels == c] == "rainy").mean()) for c in range(k)]
    sizes = [int((labels == c).sum()) for c in range(k)]
    print(f"k={k}: group sizes {sizes}, share of rainy days per group "
          f"{[round(s, 2) for s in shares]}")
````

### Diagnosis

All four are convincing. Every one of them produces groups that separate cleanly on the picture,
have sensible sizes, and could be named in a meeting without anybody objecting.

At `k=2` you would report *"two kinds of day: quiet and busy"* - and the split only half
corresponds to the weather. At `k=4` and `k=5` some groups come out entirely rainy or entirely sunny, so you would
report *"we discovered that weather defines distinct day types"* and be closer to the truth,
though still with more groups than there are weather types.

**The mechanism.** k-means does not decide how many groups exist. It is *told*, and then it
partitions the data into that many pieces as tidily as it can. It will do this for any k, on any
data, including data with no group structure whatsoever. The output is not evidence that groups
exist. It is a partition you requested.

**Why this is worse than an ordinary modelling error.** A supervised model that is wrong gets
caught, because there is a held-out score and a baseline. A clustering that is wrong produces the
same confident output as one that is right. There is no error to inspect. The check has to come
from outside the algorithm - and most of the time, nobody performs it.

### Remedies, and what each costs

| Remedy | What it fixes | What it costs | When to prefer it |
|---|---|---|---|
| Ask whether groups are *needed*, or a number would do | The whole failure | Nothing; often the answer is a regression | Always ask first |
| Check stability: re-cluster on resampled data | Groups that dissolve under resampling | A loop and some patience | Always, before naming anything (08-06) |
| Vary k and report the range | Pretending one k is the truth | Honesty in the slide | Whenever k was chosen, not measured |
| Validate against something external | Whether the groups mean anything | Needs an outside variable or an expert | Whenever a decision depends on it |
| Treat clusters as hypotheses, not findings | Overclaiming | Discipline in wording | Always |

The wording remedy sounds soft and matters most. *"Four segments exist"* is a claim about the
world. *"Asking for four groups produces these four, and they differ mainly in X"* is a
description of what you did. Only the second one is true.

This is the whole of chapter 08-06, and you have now met the reason it exists.

## A second axis: batch or online

Maria's last question - *rebuild nightly, or update with every rental?* - is not about the family
of learning at all. It is about **when the rule is updated**, and it applies to any of the
families above.

| | **Batch learning** | **Online (incremental) learning** |
|---|---|---|
| How it learns | Fit on the whole dataset at once, periodically | Update with each new observation or small batch |
| Fits in memory? | Needs the data to fit, or a strategy for it | Sees each row once, memory stays flat |
| Reacting to change | As fast as your retraining schedule | Immediately |
| Risk | Goes stale between retrains | Chases noise; one bad day of data corrupts it |
| Debugging | Easy: a fixed artefact you can re-score | Hard: the model is never the same twice |
| Typical use | Most business problems, including Maria's | High-volume streams, fast-changing behaviour, ad ranking |

The right default for almost every project is **batch**, retrained on a schedule, because it is
enormously easier to test, reproduce and roll back. Choose online when the world genuinely moves
faster than your retraining schedule can follow - and know that you have taken on a much harder
operational problem. Chapter 12-02 builds one, and 13-07 covers deploying both.

A related pair of words you will meet: **batch inference** (score a million rows overnight into
a table) versus **online inference** (answer one request in 50 milliseconds). That is a different
decision from batch versus online *learning*, and confusing the two is a common interview
stumble. 13-07 separates them.

## The map of the work

Here is the lifecycle this course follows, with the module that teaches each step.

```
 1. Frame the problem            is ML right at all?  what decision changes?   00-02, 04-01
 2. Define the unit and target   what is one row? what exactly are we predicting?  02-01, 04-01
 3. Understand the data          provenance, quality, distributions, EDA      module 02
 4. Split                        train / validation / test - correctly        04-03, 04-04
 5. Baseline                     the number every model must beat             04-02
 6. Preprocess                   impute, encode, scale - inside a pipeline    04-06, 04-07
 7. Train                        choose a family, fit it                      modules 05, 06, 10
 8. Evaluate                     the metric that matches the cost             05-04, 06-05
 9. Inspect errors               where does it fail, and on whom?             07-05
10. Interpret                    what did it use, and does that make sense?   07-06, 07-07
11. Decide the next experiment   the loop, not the end                        07-02, 07-03
12. Package, deploy, monitor     drift, retraining, rollback                  module 13
```

Two honest observations about this map.

**Step 7 is the small one.** In most real projects, framing, data understanding, splitting and
error analysis take the overwhelming majority of the effort, and fitting the model is an
afternoon. This is the opposite of how courses usually allocate their pages, which is why
modules 02 and 04 in this course are large and come before any model.

**It is a loop, not a line.** Step 9 usually sends you back to step 2 or 3, because the errors
reveal that the target was defined badly or a column means something other than you assumed.
Expect to go round several times; a project that went straight through is usually one where
nobody looked hard at the errors.

## Common misconceptions

**"Unsupervised learning is supervised learning without the effort of labelling."**
It answers a different question. Supervised: *what will happen?* Unsupervised: *how is this data
arranged?* If you need the first, no amount of clustering supplies it.

**"The clustering found four customer segments."**
It produced four groups because it was asked for four. Whether the world contains four kinds of
customer is a separate question that the algorithm did not address and cannot.

**"Self-supervised learning is unsupervised learning."**
They are often lumped together because neither needs a human labeller, but self-supervision has
a real prediction target with a real error to minimise - it was manufactured from the data
instead of provided. That difference is why it scales the way it does.

**"Reinforcement learning is just a better kind of supervised learning."**
In supervised learning your predictions do not change the data you are given. In reinforcement
learning your actions determine what you observe next, which creates the explore-exploit problem
and makes evaluation much harder. Different problem, not a stronger tool.

**"Online learning is more modern, so it is better."**
It is harder to test, harder to reproduce, harder to roll back, and it can be corrupted by a
single bad day of input. It is the right choice when change outpaces your retraining schedule,
and the wrong one otherwise.

**"Transfer learning means the model already knows my problem."**
It means the model has useful general structure. The parts specific to your problem still have
to be learned from your data, and if your data differs from the source in an important way, the
inherited structure can actively mislead. 11-05.

| Self-supervised learning | Manufacture a label by hiding part of the data and predicting it |
| What is self-supervision really for? | The representation the model builds, which transfers to the task you care about |
| Transfer learning | Start from a model trained on a large related problem, adapt with your small data |
| Reinforcement learning | Learn a policy from rewards, where your actions change the data you next observe |
| Batch vs online learning | Refit periodically on a dataset vs update continuously per observation |
| Which lifecycle step takes longest? | Framing, data understanding and error analysis - not fitting |
